## Check that PyTorch is installed and using the GPU


```python
import torch
import torchvision
from torchvision.models import resnet18, ResNet18_Weights

print(f"PyTorch version: {torch.__version__}")
print(f"Torchvision version: {torchvision.__version__}")

# 1. Check if PyTorch can see the GPU
gpu_available = torch.cuda.is_available()
print(f"Is GPU available? {gpu_available}")

if gpu_available:
    print(f"GPU Device Name: {torch.cuda.get_device_name(0)}")
    
    # 2. Test a Torchvision model on the GPU
    try:
        # Move a sample torchvision model to the GPU
        model = resnet18(weights=ResNet18_Weights.DEFAULT).cuda()
        
        # Create a dummy image tensor directly on the GPU
        dummy_input = torch.randn(1, 3, 224, 224).cuda()
        
        # Forward pass
        output = model(dummy_input)
        
        print("✅ Success: Torchvision is installed and successfully running operations on the GPU!")
    except Exception as e:
        print(f"❌ Error: PyTorch sees the GPU, but Torchvision failed to use it. Reason: {e}")
else:
    print("❌ GPU is not available. PyTorch and Torchvision are defaulting to the CPU.")
```
